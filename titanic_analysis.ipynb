```python
import urllib.request
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
urllib.request.urlretrieve(url, "titanic.csv")
print("File downloaded successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

df.head()
```

```python

print(df['Survived'].value_counts())
print("\nSurvival rate:", df['Survived'].mean())
```

```python
df.isnull().sum()
```

```python
gender_survival = df.groupby('Sex')['Survived'].mean()
print(gender_survival)

gender_survival.plot(kind='bar', color=['pink', 'lightblue'])
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.show()
```

```python
class_survival = df.groupby('Pclass')['Survived'].mean()
print(class_survival)

class_survival.plot(kind='bar', color=['gold', 'silver', '#cd7f32'])
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.xticks([0,1,2], ['1st Class', '2nd Class', '3rd Class'], rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10,6))

df[df['Survived']==0]['Age'].hist(alpha=0.5, label='Died', bins=20, color='red')

df[df['Survived']==1]['Age'].hist(alpha=0.5, label='Survived', bins=20, color='green')

plt.legend()
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
print("=" * 50)
print("TITANIC SURVIVAL ANALYSIS - KEY FINDINGS")
print("=" * 50)
print(f"1. Overall survival rate: {df['Survived'].mean()*100:.1f}%")
print(f"2. Female survival rate: {df[df['Sex']=='female']['Survived'].mean()*100:.1f}%")
print(f"3. Male survival rate: {df[df['Sex']=='male']['Survived'].mean()*100:.1f}%")
print(f"4. First class survival: {df[df['Pclass']==1]['Survived'].mean()*100:.1f}%")
print(f"5. Third class survival: {df[df['Pclass']==3]['Survived'].mean()*100:.1f}%")
print("=" * 50)
```
